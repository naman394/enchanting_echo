## Handling City's Missing Value

```python
!pip install reverse_geocoder
import reverse_geocoder as rg
import numpy as np

# Extract numeric lat/lon
df[['Lat','Lon']] = df['Coordinates'].str.split(',', expand=True)
df[['Lat','Lon']] = df[['Lat','Lon']].astype(float)

# Find missing city rows
mask = df['City'].isna()
coords = list(zip(df.loc[mask, 'Lat'], df.loc[mask, 'Lon']))

# Perform fast offline reverse geocoding
results = rg.search(coords)  # uses offline geonames database

# Fill missing cities
df.loc[mask, 'City'] = [r['name'] for r in results]

```

```python
missing_values_count = df.isnull().sum()
print(missing_values_count)
```
